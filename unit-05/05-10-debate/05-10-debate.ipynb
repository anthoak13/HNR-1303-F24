# Unit 5 Debate

## Overview

The Unit 5 group assessment is a live debate on Wednesday, December 2 (Day 29). You will be assigned to a team of five next Monday. Two teams will debate Resolution A and the other two will debate Resolution B.

Your team will be assigned to a resolution and side. You may be given a side you personally disagree with and your job is to make the best case for it. This document reflects my current plan for the debate. As the rubric and rest of the documents get written, some details may change.

## Resolutions

In a traditional debate, the statement you are debating is called a **resolution.** The affirmative side argues that it is true while the negative side argues that it is false.

### Resolution A

**An emotionally engaging chatbot can be created responsibly.**



### Resolution B

**Better testing and safeguards can prevent incidents like the Hugging Face hack.**



## Roles

Each team member will take one of the following roles. It is your job, as a team, to self-assign roles.

| Role | What you do |
| --- | --- |
| Opening | State your team’s position and define the key terms (words in the resolution that your argument depends on and that could be understood in more than one way). |
| First argument | Present your team's first main argument and its evidence |
| Second argument | Present your team's second main argument and its evidence |
| Cross-examiner | Question the other team's opening and argument speakers |
| Closer | Answer the other team's case and sum up your own |

### Rules for the opening

- You may speak from notes, but not read from a script. Notes are limited to one page of bullet points.

### Rules for arguments

- Each main argument needs a basis in both our assigned readings (including notebooks) and an outside source. Both of these should be cited in the presentation of your argument.
- You may speak from notes, but not read from a script. Notes are limited to one page of bullet points. The only full sentences allowed are direct quotations from your sources.

### Rules for cross-examination

- The cross-examiner asks questions only. A short setup is fine (referring to a specific statement made already, etc.), but no arguments can be introduced or expanded on.
- The cross-examiner asks six questions in total, with at least one to each of the other team's first three speakers.
- A speaker who is asked a question answers it personally. Teammates may not answer for them.
- Each answer is limited to 30 seconds.

### Rules for closing

- The closer must respond to at least two specific arguments the other team made, and to at least one point from the cross-examination.
- The closing may not introduce new arguments.

## Order and timing

Each debate takes about 40 minutes. We will flip a coin at the start of the debate to decide which team goes first (and therefore closes last).

| Step | Who speaks | Time |
| --- | --- | --- |
| 1. Opening | First team, then second team | 2 min each |
| 2. First argument | First team, then second team | 2.5 min each |
| 3. Second argument | First team, then second team | 2.5 min each |
| 4. Cross-examination | Second team's examiner questions the first team, then the first team's examiner questions the second team | 6 questions each, 30 sec per answer |
| 5. Preparation break | Both teams confer | 3 min |
| 6. Closing | Second team, then first team | 3 min each |
| 7. Instructor questions | I ask each of you one question about your part of the debate | 30 sec per answer |
