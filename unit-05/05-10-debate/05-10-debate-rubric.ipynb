# Unit 5 Debate Grading Scheme

The debate has five components.

| Component | Weight | Graded |
|---|---|---|
| Debate brief | 10% | As a group |
| The team's case | 40% | As a group |
| Group contribution | 10% | Individually |
| Prep document | 15% | Individually |
| Your role in the debate | 25% | Individually |

Each component is graded holistically using the rubrics below.

## Debate Brief

A single grade will be assigned to the whole team.

**A (Excellent)**

Submitted on time with every part filled in: all five roles assigned, two arguments each with an in-class source, and key terms each with a working definition. The arguments support your side of the resolution, and a teammate could plan around the brief as written.

**B (Good)**

Complete and on time, but part of it is too vague to plan around. For example, the two arguments are really the same argument, or a definition only repeats the term.

**C (Average)**

Submitted on time with one part missing, such as an argument with no source or terms listed without definitions.

**D (Poor)**

Several parts are missing, or the brief is late.

**F (Failing)**

The brief is mostly empty or not submitted.

## The Team's Case

A single grade will be assigned to the whole team. Grade levels are set mainly by the strength of your team's arguments and how persuasively you make them, across the entire debate  (opening to closing). The audience will be surveyed on the resolution before and after the debate, and how far your team moved them is a smaller part of this grade.

**A (Excellent)**

The five parts add up to one case. The opening's definitions are the ones the arguments go on to use. The two arguments are different from each other, each supports your side, and each rests on a class source and an outside source. The cross-examination goes after arguments the other team actually made, and your own speakers hold up when they are questioned. The closing statement answers the other team's case, uses what came out of the cross-examination, and sums up the case your team actually made. The case is made to the audience as well as to the other team, and it gives someone who started out undecided good reason to agree with you.

**B (Good)**

Each part is solid, but the parts do not build on each other, or the team engages only partly with what the other team said. For example, the cross-examination questions were clearly written in advance and do not fit what was said, or the closing could have been written before the debate started. Alternatively, the case is sound but does little to persuade anyone who did not already agree.

**C (Average)**

The case has a noticeable gap. An argument is thinly supported or repeats the other one, parts of the case work against each other (for example, an argument uses a term differently from the opening), or the other team's main arguments go unanswered.

**D (Poor)**

There is no recognizable team case, only five separate speakers. Little evidence is offered, and the other team's case is mostly ignored.

**F (Failing)**

The team does not present a case.

## Group Contribution

This grade is assigned to each student individually. The evidence is the survey each team member completes about how the team worked, your prep document, and what I see during the in-class prep days. You will not be penalized for a teammate's failure to participate.

**A (Excellent)**

Took part in writing the brief and in the Monday prep. Shared a prep document that teammates could use. Teammates' survey responses describe meaningful contributions, and your own account of your contributions is specific and consistent with theirs.

**B (Good)**

Contributed meaningfully, with minor gaps. For example, the prep document was hard for teammates to use, or your account differs from your teammates' in minor ways.

**C (Average)**

Participated, but teammates report noticeable gaps, such as missing a prep day or arriving after break without your part prepared, or the accounts disagree in significant ways.

**D (Poor)**

Teammates report minimal participation, and your prep document and your part in the debate support that report.

**F (Failing)**

Did not work with the team or made no meaningful contribution to its case.

## Prep Document

This grade is assigned to each student individually. What belongs in the document depends on your role:

| Role | Your work over break |
| --- | --- |
| Opening | Your team's position and its definitions of the key terms. Also how you expect the other team to define those terms, and why your team's definitions are better. |
| First and second argument | Your argument worked out in full, with its class source and its outside source. Also the questions you expect in cross-examination and how you would answer them. |
| Cross-examiner | The other team's strongest arguments as you expect them, and the holes you plan to poke in each. |
| Closer | The weak points in your own team's case, and how you would defend them. |

**A (Excellent)**

Everything listed for your role is there and worked out in detail. Claims are tied to particular things in the sources or the resolution rather than stated in general terms. What you expect from the other team is the strongest version of what they could say, and your responses deal with that version.

**B (Good)**

Everything listed for your role is there, but parts stay general, or what you expect from the other team is easy to knock down.

**C (Average)**

A required piece is missing, such as the outside source or your answers to the questions you expect. Alternatively, the document restates the team's brief without developing it.

**D (Poor)**

Little work beyond the brief is shown, or the document is late.

**F (Failing)**

Not submitted.

## Your Role in the Debate

This grade is assigned to each student individually. It is based on your speaking part, your answers in cross-examination if you are questioned, and your answer to my question at the end. What I am looking for depends on your role:

| Role | What I am looking for |
| --- | --- |
| Opening | You state your team's position clearly. You define the key terms, and the definitions are the ones your team's arguments go on to use. When you are cross-examined, you answer the question that was asked. |
| First and second argument | You make one clear claim and reason your way to it. You cite a class source and an outside source, and each one supports the claim you attach it to. When you are cross-examined, you answer the question that was asked. |
| Cross-examiner | You ask six questions, with at least one to each of the other team's first three speakers. Your questions point at specific things those speakers said, and they expose a weakness rather than just asking for a repeat. You ask questions only. |
| Closer | You respond to at least two specific arguments the other team made and to at least one point from the cross-examination. You sum up your own team's case. You introduce no new arguments. |

**A (Excellent)**

Does everything listed for the role, and responds to what actually happened in this debate rather than only delivering what was prepared. Speaks from notes rather than reading, and uses the time without running over. The answer to my question shows an understanding of their part of the case that goes beyond what they said during the debate.

**B (Good)**

Does everything listed for the role, but mostly as prepared, missing chances to respond to what the other team actually said. May fall back on reading in places, or finish well short of the time. Answers my question, but mainly by repeating what they said during the debate.

**C (Average)**

One of the things listed for the role is missing or goes wrong: for example, an argument without an outside source, a closing that answers only one of the other team's arguments, or cross-examination questions that turn into arguments. Alternatively, most of the part is read aloud. The answer to my question is on the topic but does not answer the question.

**D (Poor)**

Most of what is listed for the role is missing, or the part is read from a script. Cannot answer my question.

**F (Failing)**

Does not take part in the debate.
