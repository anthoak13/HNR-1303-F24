Let's return to our first example:

In the example above, `x` and `y` are **variables**. You can think of them as little "boxes" that store some piece of information, usually a number or a vector in our case. We can do more than *just* assign fixed numbers to a variable. We can also do arithmetic and assign the results of a computation to a variable.

````python
x = 5
y = 10

sum = x + y # sum now contains the sum of x and y
prod = x * y # prod now contains the product of x and y
diff = x - y # diff now contains the difference between x and y
frac = x / y # frac now contains the fraction of x over y expressed as a decimal
print(sum)
print(prod)
print(diff)
print(frac)
````

Very often we want to update a variable based on its *own* current value, like increasing a counter by one. Python has a shorthand for this: `x += 1` means exactly the same thing as `x = x + 1`. Python reads it as "take the current value of `x`, add 1, and store the result back in `x`." The same shorthand works for the other operators: `x -= 2`, `x *= 3`, and `x /= 4`.

**Predict** what the code below will print, then run it to check.

````python
count = 0
count = count + 1 # the long way
print(count)

count += 1 # the shorthand, does the same thing
print(count)
````

## Loops



It's incredibly common that we will want to perform the same operation over and over again (like when using the iterative method). In comes the `while` loop. A `while` loop always has the same format. It says "Do the operations in the loop until a statement is false." In code we would write this as

```python
while statement:
    Do thing 1
    Do thing 2
    ...
```

Whatever we write in `statement` will be checked by the code. If it is true, then it will run all the steps inside the body of the loop (this is the lines that are indented). Then it will check `statement` again. If it is still true, it will run the body a second time. This repeats forever until the statement is false.

The `statement` is called a **condition**, and it is always something that is either true or false, like `number < 5` ("is `number` less than 5?"). We'll look at conditions more closely later in this notebook.

Let's look at an example that prints increasing numbers.

````python
number = 0

while number < 5: # continue while number is less than 5
    print(number)
    number += 1 # increase number by one
````

That code above is short hand for the following

````python
number = 0

print(number)
number += 1
print(number)
number += 1
print(number)
number += 1
print(number)
number += 1
print(number)
number += 1


print("We will not run another time because number is now ", number, "which is not < 5")
````


Using a `while` loop is much easier to maintain than copying and pasting the code a bunch of times.  Say we decided we no longer want to print out `number` each time. Instead we want to print out 2 times `number`. In the while loop that's an easy change, and it does not depend *at all* how long the list is.



````python
number = 0

while number < 5: # continue while number is less than 5
    print(2*number)
    number += 1 # increase number by one
````

We can also add more interesting conditions like only stop when an object has hit the ground, where we might *not* know how many times through the loop it will take to get there.

## Scope

We need to now talk about **scope** or what is inside the loop and what is outside the loop. Python uses **whitespace** to denote scope. Notice in all of the loop examples above, the code running in the loop is indented with respect to the `while` loop. This means it is *within* the loop's scope. When the loop is executed only the indented code is part of the loop. The loop ends at the first line that is at the same indent level as the `while` loop.

Consider the following code. **Predict** what will be printed when you run the code block.

````python
side = 1
while side < 4:
    square = side*side
    print("side: ", side)
    side += 1

print("square:", square)
print("done")
````

The `print` statement that is indented is printed once each iteration of the loop, while the final print statement (not indented) only runs once.

# Conditional statements

We are missing one final crucial piece of information. How do we make decisions like when to stop running the `while` loop?

We make decisions based on **conditions**, a condition is a statement that is either true or false. For example, one condition might be `num == 1`. There are three parts to this condition:

1. `==`: This is the **equality operator**. It asks the question, "is the left hand side equal to the right hand side?"
1. `num`: This is the left hand side.
1. `1`: this is the right hand side.

If the value in `num` is `1` then the statement is true. Otherwise it is false.

````python
num = 1
print(num == 1)

num = 2
print(num == 1)
````

The example above also highlights the difference between `=` and `==`. The **assignment** operator (`=`) means "assign the right hand side to the variable on the left hand side." The **equality** operator (`==`) asks the question "are the two sides equal?". `==` is a question. `=` is a statement.

There are a number of **comparison operators** that we can use. Here is a short list of the operators and the questions they ask:
1. Equality: `a == b` "is `a` equal to `b`?"
1. Inequality: `a != b` "is `a` NOT equal to `b`?"
1. Greater than: ` a > b` "is `a` greater than `b`?"
1. Greater than or equal to: `a >= b` "is `a` greater than or equal to `b`?"
1. Less than: `a < b` "is `a` less than `b`?"
1. Less than or equal to: `a <= b` "is `a` less than or equal to `b`?"

We can make decisions based on these conditions using a conditional statement, also known as an `if` statement. In English a conditional statement says "if this condition is true, then execute this code. Otherwise, execute this other code."

They are structured like this:
```python
if (condition):
    #code block to execute if condition is TRUE

else:
    #code block to execute if condition is FALSE
```

Notice we have the same notion of scope we had for the `while` loop. The block of code that is executed if the condition is true is indented. That block of code ends at the `else:` line which is not indented. Let's look back at our first example:

````python
num = 1

if num == 1:
    print("num is 1")
else:
    print("num is not 1")
````

# Putting it together: `if` statements inside loops

We can put an `if` statement *inside* a `while` loop. Then the decision gets made again every time through the loop. This is how we will do things like "each time step, check whether the ball has hit the ground."

When we do this, the code has **two levels of indentation**:
- Lines indented once are inside the loop. They run every time through the loop.
- Lines indented twice are inside the `if` (or `else`). They run only when that condition is true (or false), and they get checked again every time through the loop.

**Predict** what the code below will print before you run it.

````python
number = 1

while number <= 8:
    if number > 5:
        print(number, "is big")      # inside the loop AND inside the if
    else:
        print(number, "is small")    # inside the loop AND inside the else
    number += 1                      # inside the loop only: runs every time

print("done")                        # outside the loop: runs once
````

Notice how the indentation tells you what belongs where:
- `number += 1` lines up with `if`, not with the `print` statements. It is part of the loop but not part of the `if`/`else`, so it runs every time through the loop no matter which branch was taken.
- `print("done")` is not indented at all, so it runs once after the loop finishes.
