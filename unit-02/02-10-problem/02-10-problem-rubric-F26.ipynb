# Unit 2 Assessment Grading Scheme

The Unit 2 assessment has four components.

| Component | Weight | Graded |
|---|---|---|
| Individual notebook | 50% | Individually |
| Group project notebook | 20% | As a group |
| Group contribution | 10% | Individually |
| 1:1 meeting | 20% | Individually |

The individual notebook will be graded separately and you can find your grade on that notebook. The other three components are graded holistically using the rubrics below.

## Group Project Notebook

Grade levels are set mainly by the quality of the group's reasoning and writing. Where code does not work, a clear and correct description of how it should work may take its place, earning substantial but not full credit. Citations that are missing, or cite sources which do not contain the material, can lower the grade by one or more levels. 

**A (Excellent)**

All code works and produces reasonable results. Predictions are clearly reasoned, and the updated prediction builds on the no-switch result. The plan for the switching game is complete and easy to follow, and the code carries it out, with any changes from the plan explained. The number of games simulated is justified. The final summary clearly explains the problem, the model, and the results, compares the results to both predictions, and gives a reasoned statement of confidence. Writing is detailed, logical, and easy to follow. Sources are cited where required.

**B (Good)**

Reasoning and writing meet the A standard, but one or more pieces of code do not work and are replaced by clear, correct descriptions of how they should work. Alternatively, all code works, but the reasoning lacks depth in places: predictions or confidence are only partly justified, the summary does not fully connect the results to the predictions, or the code departs from the plan in minor ways without explanation. Writing is generally clear, with minor gaps.

**C (Average)**

Every exercise is addressed with code or a description, but explanations are incomplete or unclear in several places. Descriptions of how code should work are vague or partly incorrect. Code, where present, follows a reasonable approach even if the results are wrong. The code and the plan diverge noticeably without explanation. Predictions are stated with little reasoning, and the summary reports results with little comparison or interpretation. Alternatively, the code works but the notebook misses explanation or analysis.

**D (Poor)**

One or more exercises are missing, or explanations are minimal. Code and descriptions do not follow a coherent approach. The code bears little relation to the plan. There is little connection between the predictions, the model, and the results. Writing is hard to follow. 

**F (Failing)**

The notebook is largely incomplete, with no meaningful work beyond the provided starter code and no meaningful written reasoning.

## Group Contribution

This grade is assigned to each student individually. The evidence is the group's Exercise 10 response and each partner's account in the 1:1 of their contributions and how the group worked. Where the partners' accounts conflict, each student's demonstrated understanding in the 1:1 is used to weigh them. A student is not penalized for a partner's failure to participate.

**A (Excellent)**

Worked on the project with their assigned partner, following the working-together instructions. Made meaningful contributions to the project. Their account of their own contributions and of how the group worked is specific and consistent with their partner's account and the Exercise 10 response.

**B (Good)**

Worked with their assigned partner and contributed meaningfully, but their account is vague in places or differs from their partner's in minor ways. May have occasionally departed from the working-together instructions, for example by working on part of the notebook alone.

**C (Average)**

Participated, but their partner reports noticeable gaps, such as missed work sessions or parts done alone, or the two partners' accounts disagree in significant ways.

**D (Poor)**

Their partner reports minimal participation, and the student's 1:1 performance supports that report. Alternatively, their account of their contributions contradicts their partner's account or the Exercise 10 response.

**F (Failing)**

Did not work with their assigned partner or made no meaningful contribution to the project.

## 1:1 Meeting

This grade is assigned to each student individually, based on a 15-minute meeting about the group project. Each student is responsible for understanding the whole project, not only the parts they worked on.

**A (Excellent)**

Demonstrates a strong understanding of the whole project, not only the parts they worked on. Clearly explains the project's goals and the group's approach with minimal prompting. Describes their initial prediction and reasons clearly about how the model's results confirmed or changed it, whether or not the model produced the expected result. Justifies the number of games simulated in terms of confidence in the results and the LLN. Walks through `pickDoorAvoiding` fluently and explains why it is written the way it is. Readily identifies what would need to change to extend the model to four doors, and how.

**B (Good)**

Understands the project but struggles with some details. Explains goals and approach but may need some prompting. Describes their prediction and the model's results, but the connection between them is underdeveloped. Gives a reason for the number of games simulated but does not clearly tie it to confidence in the results or LLN. Can walk through `pickDoorAvoiding` but is unclear on some choices. Identifies most of the changes needed for four doors, possibly with prompting.

**C (Average)**

Has a basic grasp of the project but struggles to explain key aspects. Can describe what was done but not always why. Describes their prediction and results but does little to relate one to the other. The number of games simulated seems arbitrary or is poorly justified. Walks through `pickDoorAvoiding` step by step but has gaps in understanding how or why it works. Identifies some changes needed for four doors but misses important ones. Understanding is noticeably uneven across the project. Explanations need frequent prompting.

**D (Poor)**

Struggles to explain the project and how its parts fit together. Goals are unclear. Cannot meaningfully connect their prediction to the model's results or explain the number of games simulated. Explanation of `pickDoorAvoiding` is incomplete or confused. Cannot identify what would change for four doors without extensive prompting. Understanding of their own work is limited, or confined to a small part of it.

**F (Failing)**

Decodes their own code in real time with no demonstrated understanding beyond that. Cannot describe the project's goals, their prediction, or how the model relates to either. Responses lack coherence, and no clear reasoning is given.
