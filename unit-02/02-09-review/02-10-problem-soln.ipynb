## The Model - Description
As the puck falls, it bounces off pegs in the board. We will assume that after hitting a peg, the ball will either bounce half a unit of distance to the right or half a unit of distance to the left. So after two pegs it will have moved either no distance, one unit to the right, or one unit to the left (just like our mosquito). The board in the initial model is constructed so the falling ball hits 10 pegs before reaching the base of the board.

We can model the fall of each ball as a random walk, which each step in the walk represents the ball bounding off a peg in a row. We can track the position of the ball as it falls. In the model, we will drop the puck starting at the position `startingPos`. As it falls, it has a 50% chance of bouncing right and a 50% chance of bouncing left. We then update the position of the ball depending on the result. After bouncing 10 times we record the final position of the ball.

```python
## Inputs and outputs
Nball = 100 #Number of trials 
L = 1 # Distance between pegs
Nstage = 10 # Number of levels the ball will bounce down
starting_pos = 0 # Starting position of the ball

NScholarship = 0
NDiscount = 0
NFail = 0

final_pos = [] # Final positions of all the balls

# This defines a function to simulate dropping a ball down the Plinko game
# Returns the final position of the ball that fell.
def drop_ball():
    global starting_pos, Nstage
    x = starting_pos # Initial position of the ball
    for i in range(Nstage):
        dir = rand.randint(0,1)
        if dir == 0:
            x = x + L/2
        else:
            x = x - L/2
    return x

## Beginning of Monte Carlo loop
for i in range(Nball):
    
    pos = drop_ball()
    
    if pos == 0:
        NScholarship = NScholarship + 1
    elif pos == 1 or pos == -1:
        NDiscount = NDiscount + 1
    else:
        NFail = NFail + 1
    
    final_pos.append(pos)

###
# Add code to calculate the output necessary to answer our questions
#
###

## Plotting code - will display the results of the simulation in a histogram
SlotaRange = Nstage/2
plt.xlim(-SlotaRange-1,SlotaRange+1)

bins = np.arange(-SlotaRange-0.5, SlotaRange+.5, 1)
histogram = plt.hist(final_pos, bins=bins, facecolor='gray',edgecolor='black')
plt.plot([-0.5,0.5],[0,0], linewidth=10, color='purple')
plt.plot([-1.3,-0.7],[0,0], linewidth=10, color='blue')
plt.plot([0.7,1.3],[0,0], linewidth=10, color='blue')


plt.show()

```

In the game of Yahtzee, in one turn, you roll five d6s up to three times. You can earn points for the following:

- Small Straight (a sequence of 4)
- Large Straight (a sequence of 5)
- 3 of a kind
- 4 of a kind
- YAHTZEE (5 of a kind)
- Full House (3 dice are the same and 2 dice are the same)

I've defined a function `turn()` below that accepts an argument that is a string ("3 of a kind" or "4 of a kind", etc.) for what you are looking for and returns `True` if it was rolled during the turn and `False` if it was not rolled during the turn.

There is an array named `dice`. If you `print(dice)` you can see the final roll of the dice, with the dice arranged in ascending order.

Run the cell below to define the `turn()` function and `dice` array.

```python
dice = np.zeros(5) # define an array for the dice

def turn(s):
    # s can be any of the following:
    # "3 of a kind"
    # "4 of a kind"
    # "Full House"
    # "Small Straight"
    # "Large Straight"
    # "YAHTZEE"
    
    result = 0 #default is 0
    global dice

    for i in range(3):
        dice = np.random.randint(1,7, size=5)
        dice.sort()

        if s == "YAHTZEE":
            if dice[0] == dice[1] == dice[2] == dice[3] == dice[4]:
                return True

        elif s == "Large Straight":
            u = np.unique(dice).tolist()
            if( u == [1,2,3,4,5] or u == [2,3,4,5,6]):
                return True

        elif s == "Small Straight":
            u = np.unique(dice)
            if len(u) >= 4:
                if u[0] == 1 and u[1] == 2 and u[2] == 3 and u[3] == 4:
                    return True
                elif u[0] == 2 and u[1] == 3 and u[2] == 4 and u[3] == 5:
                    return True
                elif u[0] == 3 and u[1] == 4 and u[2] == 5 and u[3] == 6:
                    return True

        elif s == "4 of a kind":
            u, counts = np.unique(dice, return_counts=True)
            if np.any(counts == 4):
                return True

        elif s == "3 of a kind":
            u, counts = np.unique(dice, return_counts=True)
            if np.any(counts == 3):
                return True

        elif s == "Full House":
            u, counts = np.unique(dice, return_counts=True)
            if np.any(counts == 3) and np.any(counts == 2):
                return True

    return False
```
