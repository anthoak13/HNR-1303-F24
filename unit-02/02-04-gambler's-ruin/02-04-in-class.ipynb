## Casino

1. What is the question we are trying to answer with this model?
- How much money does the player make for x number of bets?
1. What are our model inputs?
- Number of plays, amount of money bet, house edge.
1. What are our model outputs?
- change in player money
1. What constitutes a single "trial" in our model?
- One bet
1. What information do we need to save from each "trial"?
- win/loss (used to update player money)
1. How do we use our model outputs to answer the question identified above?
- Compare casino profit (change in player oney) to number of bets.


## Coins/Dice

1. What is the question we are trying to answer with this model?
- Probability of tossing heads? Probability of two die summing to 7? Probability of rolling doubles?
1. What are our model inputs?
- probability of tossing a head
- Number that we're summing to
- even die
1. What are our model outputs?
- Number of sucesses
1. What constitutes a single "trial" in our model?
- One coin toss or one dice roll.
1. What information do we need to save from each "trial"?
- If it was successful
1. How do we use our model outputs to answer the question identified above?
- Percentage (number success)/(number of trials)

## Dart Fundraiser

1. What is the question we are trying to answer with this model?
- How much is the expected profit per dart
1. What are our model inputs?
- radius of circle, payout per circle, price per dart
1. What are our model outputs?
- Number landed in each circle, prob in each circle
1. What constitutes a single "trial" in our model?
- One dart throw
1. What information do we need to save from each "trial"?
- Location on board(drawing picture), and which circle it ended up in.
1. How do we use our model outputs to answer the question identified above?
- Use probability: `(trials*cost -  success*payout)/trials`
