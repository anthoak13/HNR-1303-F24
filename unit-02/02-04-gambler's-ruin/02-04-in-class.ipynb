## Random Walk

1. What is the question we are trying to answer with this model?
- What are the probabilities a dead mosquito is found some distance from its birthpplace?
1. What are our model inputs?
-  Number of mosquitos, number of stage, length of a step
1. What are our model outputs?
- Probability of each distance (where each mosquito landed)
1. What constitutes a single "trial" in our model?
- Life of one mosquito
1. What information do we need to save from each "trial"?
- Distance from breeding pool (number at 0, 1, 2, etc)
1. How do we use our model outputs to answer the question identified above?
- To get prob we divide number at location by total number of mosquitos
