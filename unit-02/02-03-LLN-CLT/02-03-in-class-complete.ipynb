## Law of Large Numbers and central limit theorem

1. Three points of Central Limit Theorem (CLT)
- We have a random procees gives outcome
- If we add a bunch of those outcomes together we get a distrobution
- As the number we add together increases, the distro looks more like a bell curve

2. Connection of CLT to MC models
- We get an random outcome: each trial gives us random number. For area 1/0
- Add a bunch together -> gives us area.
- CLT tells us how close we *expect* our area to be (predict std. dev.)

3. Three assumptions underlying CLT
- Samples must be independent
- Samples must draw from same distribution
- Variance cannot be infinite

4. Example of MC that might violate assumptions
- Galton board simulation if balls could collide then they would not be independent

5. What does Law of Large Numbers promise? What does CLT give us in addition?
- LLN tellls us that the mean we get with averaging all our trials -> true mean. Also tell us how the standard deviation changes. CLT tell us that the shape goes to a bell curve

```python
import random as rand #for random numbers
import numpy as np

## Start of inputs
P = 5 #price per dart
Ntosses = 100 #number of darts thrown

L = 45 #the length of one side of the dart board in cm
Rpurple = 45/2
Rblue = 27/2
Rred = 9/2

Ppurple = 1
Pblue = 5
Pred = 10
## End of inputs

## Start of outputs (updated each trial)
Npurple = 0
Nblue = 0
Nred = 0

xmodel = []
ymodel = []
## End of outputs

## Run many trials
for i in range(1,Ntosses+1):

    ## Start of single trial
    x = rand.uniform(-L/2, L/2) #should be a random number
    y = rand.uniform(-L/2, L/2) #should be a random number
    ## End of single trial


    ## Start evaluating results of trial and saving result
    xmodel.append(x)
    ymodel.append(y)

    ######
    # See if the x,y coordinates of a dart is within each circle and count Npurple, Nblue, or Nred
    d = np.sqrt(x**2+y**2)
    if d < Rred:
      Nred += 1
    if d > Rred and d < Rblue: # could also do: elif d < Rblue
      Nblue += 1
    if d > Rblue and d < Rpurple: # could also do: elif d < Rpurple
      Npurple += 1
    ######

    ## End evaluating results of trial and saving result

## Start calculating output to study
gross = P*Ntosses
loss = Ppurple*Npurple + Pblue*Nblue + Pred*Nred
## End calcualting output to study

## Start reporting results
print("Percent in red = {:.2f}%".format(Nred/Ntosses*100))
print("Percent in blue = {:.2f}%".format(Nblue/Ntosses*100))
print("Percent in purple = {:.2f}%".format(Npurple/Ntosses*100))
print("Percent not in a circle = {:.2f}%".format((Ntosses-Npurple-Nblue-Nred)/Ntosses*100))
print("Gross = ${:.2f}".format(gross))
print("Loss = ${:.2f}".format(loss))
print("Profit = ${:.2f} = {:.2f}% of gross".format(gross-loss, (gross-loss)/gross*100))
## End reporting results
```
