# Unit 4 Assessment Grading Scheme

The Unit 4 assessment has four components.

| Component | Weight | Graded |
|---|---|---|
| Individual notebook | 50% | Individually |
| Group project notebook | 20% | As a group |
| Group contribution | 10% | Individually |
| 1:1 meeting | 20% | Individually |

The individual notebook will be graded separately and you can find your grade on that notebook. The other three components are graded holistically using the rubrics below.

## Group Project Notebook

See the rubric specifically for the group notebook.

## Group Contribution

This grade is assigned to each student individually and is based on both partners' accounts in the 1:1. You will not be penalized for a partner's failure to participate.

**A (Excellent)**

Worked on the project with their assigned partner, following the working-together instructions. Made meaningful contributions to the project. Their account of their own contributions and of how the group worked is specific and consistent with their partner's account.

**B (Good)**

Worked with their assigned partner and contributed meaningfully, but their account is vague in places or differs from their partner's in minor ways. May have occasionally departed from the working-together instructions, for example by working on part of the notebook alone.

**C (Average)**

Participated, but their partner reports noticeable gaps, such as missed work sessions or parts done alone, or the two partners' accounts disagree in significant ways.

**D (Poor)**

Their partner reports minimal participation, and the student's 1:1 performance supports that report. Alternatively, their account of their contributions contradicts their partner's account.

**F (Failing)**

Did not work with their assigned partner or made no meaningful contribution to the project.

## 1:1 Meeting

This grade is assigned to each student individually, based on a 15-minute meeting about the group project. Each student is responsible for understanding the whole project, not only the parts they worked on.

**A (Excellent)**

Demonstrates a strong understanding of the whole project, not only the parts they worked on. Clearly explains the project's goals and the group's approach with minimal prompting. Walks fluently through how reinfection and the second modification change the model and why. Justifies the group's verification of their modifications. Describes the group's question and reasons clearly about how their model answers it. Readily identifies what would need to change to implement the modification the group did not choose, and how.

**B (Good)**

Understands the project but struggles with some details. Explains goals and approach but may need some prompting. Can walk through how the model was changed but is unclear on some choices. Describes how the modifications were verified but does not clearly justify the verification. Describes what their model showed but is unclear on how it answers the group's question. Identifies most of the changes needed to implement the modification the group did not choose, possibly with prompting.

**C (Average)**

Has a basic grasp of the project but struggles to explain key aspects. Can describe what was done but not always why. Walks through the changes to the model step by step but has gaps in understanding how or why they work. Verification is vague or poorly justified. Describes the group's question and their model's results but does little to relate one to the other. Identifies some changes needed to implement the modification the group did not choose but misses important ones. Understanding is noticeably uneven across the project. Explanations need frequent prompting.

**D (Poor)**

Struggles to explain the project and how its parts fit together. Goals are unclear. Explanation of the changes to the model is incomplete or confused. Cannot meaningfully explain how the modifications were verified or connect the group's question to their model's results. Cannot identify what would change to implement the modification the group did not choose without extensive prompting. Understanding of their own work is limited, or confined to a small part of it.

**F (Failing)**

Decodes their own code in real time with no demonstrated understanding beyond that. Cannot describe the project's goals, the changes made to the model, or how the model relates to the group's question. Responses lack coherence, and no clear reasoning is given.
