# Original Model

The tables below summarize the SIR model we developed in class. See the `04-04-SIR-model` notebook for a more detailed description of the model.


### Populations and interactions in the model
Population | Process or Interaction | Associated Rate of Change | Proportionality Constant
--- | --- | --- | ---
Susceptibles (S) | becomes infected; depends on number of S and I interactions | $-(r_{S,I})SI$ | $r_{S,I}$
Recoverds (R) | recovers from infection; depends on number infected | $+(r_{R,I})I$ | $r_{R,I}$
Infecteds (I) | a susceptible becomes infected; the opposite of $\dot{S}$ | $-\frac{\Delta S}{\Delta t}=(r_{S,I})SI$ |
Infecteds (I) | an infected recovers; the opposite of $\dot{R}$ | $-\frac{\Delta R}{\Delta t}=-(r_{R,I})I$ |

### Parameters in the model
Model parameter | Process or Interaction | Example Value
--- | --- | --- |
$f$ | Number of interactions per day of an infected person with the entire population | 50 per day
$p_{trans}$ | The fraction of interactions with a susceptible person that results in a transmission (i.e. the probability of getting infected each interaction) | 0.05
$N$ | Total alive population | 1000
$t_{recov}$ | Average time for an infected person to recover and stop being infectious | 2 days
$r_{S,I}$ | Infection rate | $\frac{f p_{trans}}{N}$
$r_{R,I}$ | Recovery rate | $\frac{1}{t_{recov}}$

### Rate of change equations in the model

$$\begin{align*}
\dot{S} &= -(r_{S,I})SI \\
\dot{R} &= (r_{R,I})I \\
\dot{I} &= (r_{S,I})SI - (r_{R,I})I
\end{align*}
$$

**Run the following code cell**. It will define a function called `runOrigModel(f, p_trans, N, t_recov)` that takes in our four input parameters and runs the model for fifty days.

```python
import numpy as np
import matplotlib.pyplot as plt

def runOrigModel(f,p_trans, N, t_recov):
    t = 0
    dt = 0.01 #unit is day
    t_total = 50 #number of days to run the model

    S = N - 1
    I = 1
    R = 0


    r_SI = f*p_trans/N #transmission coefficient
    r_RI = 1/t_recov #recovery coefficient

    tmodel = []
    Smodel = []
    Imodel = []
    Rmodel = []

    while t < t_total:

        Srate = -r_SI*S*I
        Rrate = r_RI*I
        Irate = -Srate + -Rrate

        S = S + Srate*dt
        R = R + Rrate*dt
        I = I + Irate*dt

        t = t + dt

        tmodel.append(t)
        Smodel.append(S)
        Rmodel.append(R)
        Imodel.append(I)

    fig = plt.figure(figsize=(8,6))
    plt.title("SIR Model")
    plt.xlabel("Time (day)")
    plt.ylabel("Population")
    plt.plot(tmodel, Smodel, label='Susceptibles')
    plt.plot(tmodel, Rmodel, label='Recoverds')
    plt.plot(tmodel, Imodel, label='Infecteds')
    plt.text(0.68*t_total,500,f'$r_{{infection}}$ = {r_SI:.2e} day$^{{-1}}$')
    plt.text(0.68*t_total,450,f'$r_{{recovery}}$ = {r_RI:.2e} day$^{{-1}}$')
    plt.grid()
    plt.legend()
    plt.show()
```
