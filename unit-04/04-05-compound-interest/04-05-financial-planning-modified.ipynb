You are evaluating your retirement. At 25 years old, you invest monthly into an IRA (Individual Retirement Account) with an average 7% interest rate per year. After your retirement age, you want to live solely on that investment. What is the minimum age you can retire if you die with exactly zero dollars left in your investment?

### Assumptions

1. How much we will spend after retirement (3k/month)
1. No income post-retirement
1. Need to know age of death - 80 years
1. Contribute same amount each year - 1.6k/month
1. Frequency of contributions (yearly, monthly, etc.)
1. Assume constant intrest rate - 7%
1. Need some assumption about inflation - constant percentage (~3%)
  1. 'real rate of return' = rate - inflation. Assume salary and cost match inflation exactly
1. How often we compund interest
1. Perpetually single

### Assumptions



```python
import matplotlib.pyplot as plt
# Outline of code

# Define variables/model parameters
death_age = 80
age = 25
retirement_age = 49 # Change until no money at 80

# Salary/spending assumptions
#contrib = 1600 * 12 # per year
contrib = 1400 * 12 # per year
spending = 3000 * 12 # per year

dt = 1/12 # in years

#Market assumptions
return_rate = 0.07
inflation = 0.03
real_rate = return_rate - inflation

balance = 0

balance_list = []
age_list = []

#Loop
while age < death_age:

  interest = balance * real_rate * dt
  balance = balance + interest

  # If not retired - add contribution and intrest
  if age < retirement_age:
    balance = balance + contrib * dt
  else:
    balance = balance - spending * dt

  # if retired - subtract expenses and add intrest
  age = age + dt

  # Save in list
  balance_list.append(balance)
  age_list.append(age)

#report
print(f"At death we have ${balance} left in our retirement account.")
fig = plt.figure()
plt.title("Invesment over time")
plt.xlabel("Age (years)")
plt.ylabel("Balance ($)")
plt.plot(age_list, balance_list)
plt.show()



```
