# Financial Planning


This is an open-ended question that requires you to make assumptions, and making assumptions is an integral part of *modeling*.

You are evaluating your retirement. At 25 years old, you invest monthly into an IRA (Individual Retirement Account) with an average 7% interest rate per year. After your retirement age, you want to live solely on that investment. **What is the minimum age you can retire if you die with exactly zero dollars left in your investment?**

We will define variables as needed and use variables in all calculations. This allows us to explore predictions of the model by adjusting the values of the variables.

### Assumptions

- Invest same amount each month/year: \$600/month
- Value of initial invesment: \$1000
- How old you are when you die: 85-110 years old
- Assume some constant inflation rate: 3.1\%
- Assumed no tax (roth IRA)
- Assume constant interest rate: 7\%
- How much will we spend post-retirement in a month/year: $5000
- Only retirement account/source of income in retirement



### Assumptions



```python
import matplotlib.pyplot as plt
# Outline of code

# Define variables/model parameters
age = 25
death_age = 90
retirement_age = 56

# Salary/spending
contrib = 600 * 12 # $/year
spending = 5000 *12 # $/year

# Assumptions on interest
return_rate = 0.1 # % / year
inflation = 0.031 # % / year
real_rate = return_rate - inflation
dt = 1/12 # timestemp in years

# Initial conditions
balance = 1000

# Lists for plotting
balance_list = [balance]
age_list = [age]

# Loop
while age < death_age:

  # Calculate and add interest in into account
  interest = balance*real_rate*dt
  balance = balance + interest

  # if not retired we will contribute
  if age < retirement_age:
    balance = balance + contrib * dt
  # else we will take out money
  else:
    balance = balance - spending * dt

  # Increase time
  age = age + dt

  # Save current balance and age in lists for plotting
  age_list.append(age)
  balance_list.append(balance)

# Report
fig = plt.figure()
plt.title("IRA balance over time")
plt.xlabel('Age (years)')
plt.ylabel('Balance (US dollar)')
plt.plot(age_list, balance_list, 'b-')
plt.grid()
plt.show()


```
