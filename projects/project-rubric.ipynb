# Rubric for Final Project
**Total: 100 Points (5 categories)**  

---

### **1. Problem Selection, Research Question, and Significance (20%)**  
This section evaluates the selection of a problem, formulation of a research question, and the explanation of its significance.

- **A (Excellent):** The problem is clearly and thoughtfully selected, with a well-developed, specific, and relevant research question. The significance of the problem is thoroughly explained, including its broader context or implications.  
- **B (Good):** The problem and research question are appropriate but could be more specific or detailed. The significance is explained but may lack depth or context.  
- **C (Satisfactory):** The problem and research question are present but underdeveloped. The significance is vaguely addressed or lacks sufficient detail.  
- **D (Poor):** The problem and research question are unclear, irrelevant, or missing. The significance is not addressed or poorly articulated.  

---

### **2. Model Development, Accuracy, and Refinement (25%)**  
This section evaluates the development of a mathematical model, its numerical implementation, and refinement based on test cases or comparisons to theoretical predictions.

- **A (Excellent):** The model is fully and correctly implemented, with appropriate assumptions and mathematical techniques applied. Test cases or comparisons are used effectively to validate and refine the model, demonstrating confidence in its accuracy.  
- **B (Good):** The model is mostly implemented correctly but may have minor issues in assumptions, techniques, or validation. Test cases are present but not fully explored.  
- **C (Satisfactory):** The model is partially implemented or contains significant errors. Validation through test cases or comparisons is superficial or incomplete.  
- **D (Poor):** The model is poorly implemented or missing key components. There is little to no validation or refinement.  

---

### **3. Data Analysis and Presentation of Results (20%)**  
This section evaluates the interpretation of results and how well the analysis answers the research question.

- **A (Excellent):** Data analysis is thorough and insightful, with clear interpretation that directly addresses the research question. Results are presented clearly using well-labeled graphs, tables, and narrative explanations.  
- **B (Good):** Data analysis is sound but may lack depth or fail to explore all relevant scenarios. Results are presented clearly but could use more explanation or better visuals.  
- **C (Satisfactory):** Data analysis is incomplete or lacks depth, with results that only partially address the research question. Presentation of results is unclear or missing some visuals or explanations.  
- **D (Poor):** Data analysis is superficial or missing. Results do not address the research question, and presentation is unclear or incomplete.  

---

### **4. Communication and Presentation (20%)**  
This section evaluates the integration of narrative and code, the quality of the Jupyter notebook, and the in-class presentation.

- **A (Excellent):** The Jupyter notebook integrates code and narrative seamlessly, telling a coherent story of the problem and solution. The recorded presentation is clear, engaging, and well-organized, with effective responses to questions.  
- **B (Good):** The Jupyter notebook is clear but could better integrate narrative and code. The presentation is effective but may lack engagement or organization in places. Responses to questions are satisfactory.  
- **C (Satisfactory):** The Jupyter notebook is somewhat disorganized or lacks coherence between narrative and code. The presentation is unclear or disjointed, with incomplete responses to questions.  
- **D (Poor):** The Jupyter notebook and presentation are unclear or poorly organized, with minimal integration of code and narrative. Responses to questions are inadequate or absent.  

---

### **5. Collaboration and Individual Contribution (15%)**  
This section evaluates teamwork (for groups) or peer review and feedback (for individuals).

- **A (Excellent):** For groups: Roles are clearly defined, and contributions are evenly distributed. Team members demonstrate strong collaboration during the presentation.  
  For individuals: The notebook has been reviewed, feedback has been incorporated, and the presentation shows evidence of practice and constructive critique.  
- **B (Good):** For groups: Roles are defined, but contributions may not be evenly distributed. Collaboration during the presentation is adequate but could be improved.  
  For individuals: The notebook has been reviewed, but feedback may not be fully incorporated. Presentation practice is evident but could be improved.  
- **C (Satisfactory):** For groups: Roles are vaguely defined or unevenly distributed. Collaboration during the presentation is limited.  
  For individuals: Feedback is minimally addressed or ignored. Presentation lacks evidence of practice or critique.  
- **D (Poor):** For groups: Roles are unclear or contributions are highly uneven. Collaboration is ineffective.  
  For individuals: Feedback is not addressed, and presentation preparation is minimal.  

---

### **Total: 100 Points**
