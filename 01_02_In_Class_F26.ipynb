# Steps of Modeling Process

1. Analyze the problem
  - Study and ID actual problem/question
  - Identifying variables
  - Are there simplifying assumtions?

1. Formulate a model
  - Use assumptions and variables
  - Gather data and maping to variables
  - What is the relationship or equation relating the variables
1. Solve the model
  - Do the math
1. Verify and iterperte
  - Change different inputs and look at output (does it make sense?)
  - Using results of math/model (does it answer original question?)
  - Compare to other sources or other models
  - Liming case - simplier problem that you can solve with same model
1. Report
  - Compile everything you've done above into a single document for other
  - why? outcomes, etc.
  - Maybe suggestions for future iterations, things to explore, etc.
1. Maintain/update
