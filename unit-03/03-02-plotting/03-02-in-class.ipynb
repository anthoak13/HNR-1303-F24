## Empirical Models

A mathematical expression that follows real world data. Can be used to make predictions where the model is valid (interpolation). Does **not** explain the data or describe why the data has the trend that it does.

# Review (Functions)

## Mathematical functions

$$f(x) = a\sqrt{x}+bx+cx^2$$

What are the parts of a function?

- Input $x$, independent variable
- Output $f(x)$: dependent variable
- Constants ($a,b,c$): Model parameters

## Programming functions
How would we implement the above function in python?


```python
# Define a lambda function
f = lambda x,a,b,c: a*np.sqrt(x) + b*x + c*x**2
print(f(1,0,0,1))


# Define a normal function
def f(x,a,b,c):
  return a*np.sqrt(x) + b*x + c*x**2
print(f(1,0,0,1))
```
